```python
import pandas as pd  # Import pandas for data manipulation and analysis
import numpy as np  # Import numpy for numerical operations
import matplotlib.pyplot as plt  # Import matplotlib for data visualization

df = pd.read_csv(
    "../data/raw/Dataset.csv"
)  # Load the new phishing URL dataset

print("Dataset Loaded Successfully!")  # Confirm that the dataset loaded successfully
print("Shape:", df.shape)  # Display the number of rows and columns

print("\nFirst 5 rows:")  # Display heading
print(df.head())  # Display the first five rows

print("\nDataset Information:")  # Display heading
df.info()  # Display columns, data types and non-null values

print("\nMissing Values:")  # Display heading
print(df.isnull().sum())  # Count missing values in every column

print("\nDuplicate Rows:")  # Display heading
print("Duplicate rows:", df.duplicated().sum())  # Count completely duplicate rows

print("\nStatistical Summary:")  # Display heading
print(df.describe())  # Display statistical summary of numerical features

print("\nClass Distribution:")  # Display heading
print(df["label"].value_counts())  # Count legitimate and phishing samples

print("\nClass Distribution Percentage:")  # Display heading
print(df["label"].value_counts(normalize=True) * 100)  # Calculate percentage of each class

counts = df["label"].value_counts()  # Count each class for visualization

plt.figure(figsize=(5, 4))  # Set graph size
plt.bar(
    ["Legitimate", "Phishing"],
    counts.values
)  # Create class distribution bar chart

plt.title("Website Class Distribution")  # Add chart title
plt.xlabel("Website Type")  # Label x-axis
plt.ylabel("Number of URLs")  # Label y-axis
plt.show()  # Display the chart
```
