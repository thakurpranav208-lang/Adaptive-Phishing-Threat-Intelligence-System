```python
import pandas as pd
import numpy as np
import sys
import os

# Allow notebook to import the backend predictor
sys.path.append("../backend")

from predictor import extract_features


# Load the original dataset
df = pd.read_csv("../data/raw/Dataset.csv")

# Select one known phishing URL
phishing_row = df[df["label"] == 1].iloc[0]

url = phishing_row["url"]

print("URL:")
print(url)

print("\nActual features stored in Dataset.csv:")
actual = phishing_row.drop(
    labels=["url", "dom", "tld", "label"]
)

print(actual)


print("\nFeatures generated by our predictor.py:")
generated = extract_features(url).iloc[0]

print(generated)


print("\n================ FEATURE COMPARISON ================\n")

comparison = pd.DataFrame({
    "Dataset_Value": actual,
    "Predictor_Value": generated
})

comparison["Difference"] = (
    comparison["Predictor_Value"] -
    comparison["Dataset_Value"]
)

print(comparison.to_string())


print("\nNumber of different features:",
      (comparison["Difference"].abs() > 0.000001).sum())
```
