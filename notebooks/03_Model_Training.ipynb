```python
import pandas as pd  # Import pandas for data handling
import numpy as np  # Import numpy for numerical operations
import joblib  # Import joblib for loading and saving models

from sklearn.linear_model import LogisticRegression  # Import Logistic Regression
from sklearn.tree import DecisionTreeClassifier  # Import Decision Tree
from sklearn.ensemble import RandomForestClassifier  # Import Random Forest
from sklearn.svm import LinearSVC  # Import Linear SVM
from xgboost import XGBClassifier  # Import XGBoost


X_train_scaled = joblib.load(
    "../data/processed/X_train_scaled.pkl"
)  # Load scaled training features

X_test_scaled = joblib.load(
    "../data/processed/X_test_scaled.pkl"
)  # Load scaled testing features

y_train = joblib.load(
    "../data/processed/y_train.pkl"
)  # Load training labels

y_test = joblib.load(
    "../data/processed/y_test.pkl"
)  # Load testing labels


logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)  # Create Logistic Regression model

logistic_model.fit(
    X_train_scaled,
    y_train
)  # Train Logistic Regression


decision_tree_model = DecisionTreeClassifier(
    random_state=42
)  # Create Decision Tree model

decision_tree_model.fit(
    X_train_scaled,
    y_train
)  # Train Decision Tree


random_forest_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)  # Create Random Forest model

random_forest_model.fit(
    X_train_scaled,
    y_train
)  # Train Random Forest


svm_model = LinearSVC(
    random_state=42,
    max_iter=5000
)  # Create Linear SVM model

svm_model.fit(
    X_train_scaled,
    y_train
)  # Train Linear SVM


xgb_model = XGBClassifier(
    n_estimators=100,
    random_state=42,
    eval_metric="logloss",
    n_jobs=-1
)  # Create XGBoost model

xgb_model.fit(
    X_train_scaled,
    y_train
)  # Train XGBoost


joblib.dump(
    logistic_model,
    "../data/models/logistic_regression.pkl"
)  # Save Logistic Regression model

joblib.dump(
    decision_tree_model,
    "../data/models/decision_tree.pkl"
)  # Save Decision Tree model

joblib.dump(
    random_forest_model,
    "../data/models/random_forest.pkl"
)  # Save Random Forest model

joblib.dump(
    svm_model,
    "../data/models/svm.pkl"
)  # Save SVM model

joblib.dump(
    xgb_model,
    "../data/models/xgboost.pkl"
)  # Save XGBoost model
```
