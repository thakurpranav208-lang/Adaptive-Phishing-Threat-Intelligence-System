```python
import pandas as pd  # Import pandas for data handling
import numpy as np  # Import numpy for numerical operations
import joblib  # Import joblib for loading models

from sklearn.metrics import (
    accuracy_score,  # Calculate accuracy
    precision_score,  # Calculate precision
    recall_score,  # Calculate recall
    f1_score,  # Calculate F1-score
    confusion_matrix,  # Create confusion matrix
    classification_report  # Generate classification report
)


X_test_scaled = joblib.load(
    "../data/processed/X_test_scaled.pkl"
)  # Load scaled testing features

y_test = joblib.load(
    "../data/processed/y_test.pkl"
)  # Load testing labels


logistic_model = joblib.load(
    "../data/models/logistic_regression.pkl"
)  # Load Logistic Regression model

decision_tree_model = joblib.load(
    "../data/models/decision_tree.pkl"
)  # Load Decision Tree model

random_forest_model = joblib.load(
    "../data/models/random_forest.pkl"
)  # Load Random Forest model

svm_model = joblib.load(
    "../data/models/svm.pkl"
)  # Load SVM model

xgb_model = joblib.load(
    "../data/models/xgboost.pkl"
)  # Load XGBoost model


models = {
    "Logistic Regression": logistic_model,
    "Decision Tree": decision_tree_model,
    "Random Forest": random_forest_model,
    "SVM": svm_model,
    "XGBoost": xgb_model
}  # Store all models for comparison


results = []  # Create list to store model results


for name, model in models.items():
    y_pred = model.predict(X_test_scaled)  # Generate predictions

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1 Score": f1_score(y_test, y_pred)
    })  # Store evaluation metrics


results_df = pd.DataFrame(results)  # Convert results into a DataFrame

results_df = results_df.sort_values(
    by="F1 Score",
    ascending=False
)  # Sort models by F1 score


print(results_df.to_string(index=False))  # Display model comparison


for name, model in models.items():
    y_pred = model.predict(X_test_scaled)  # Generate predictions

    print("\n" + "=" * 60)  # Separate model results
    print(name)  # Display model name
    print("=" * 60)

    print("\nConfusion Matrix:")
    print(confusion_matrix(y_test, y_pred))  # Display confusion matrix

    print("\nClassification Report:")
    print(classification_report(y_test, y_pred))  # Display detailed metrics
```
